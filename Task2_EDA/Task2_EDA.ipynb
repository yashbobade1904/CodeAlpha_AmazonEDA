```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
print("✅ Libraries loaded")
```

## 1. Load Dataset

```python
df_raw = pd.read_csv('../amazon.csv')
print(f"Raw shape: {df_raw.shape}")
df_raw.head(3)
```

## 2. Data Cleaning & Type Conversion

```python
df = df_raw.copy()

def clean_price(col):
    return col.str.replace('₹', '', regex=False).str.replace(',', '', regex=False).str.strip().astype(float)

df['discounted_price'] = clean_price(df['discounted_price'])
df['actual_price']     = clean_price(df['actual_price'])
df['discount_pct']     = df['discount_percentage'].str.replace('%', '', regex=False).str.strip().astype(float)
df['rating']           = pd.to_numeric(df['rating'], errors='coerce')
df['rating_count']     = df['rating_count'].str.replace(',', '', regex=False).str.strip()
df['rating_count']     = pd.to_numeric(df['rating_count'], errors='coerce')
df['main_category']    = df['category'].str.split('|').str[0]
df['sub_category']     = df['category'].str.split('|').str[1].fillna('Unknown')
df['savings']          = df['actual_price'] - df['discounted_price']

df_clean = df.dropna(subset=['rating']).copy()
print(f"After cleaning: {df_clean.shape[0]} rows  (dropped {df.shape[0]-df_clean.shape[0]} invalid rows)")
df_clean[['product_name','main_category','discounted_price','actual_price','discount_pct','rating','rating_count']].head()
```

## 3. Dataset Structure

```python
print("="*50)
print("DATASET INFO")
print("="*50)
df_clean.info()
```

## 4. Descriptive Statistics

```python
df_clean[['discounted_price','actual_price','discount_pct','rating','rating_count','savings']].describe().round(2)
```

## 5. Missing Value Analysis

```python
missing = df_clean.isnull().sum()
missing_pct = (missing / len(df_clean) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
missing_df = missing_df[missing_df['Missing Count'] > 0]
if missing_df.empty:
    print("✅ No missing values in cleaned dataset!")
else:
    print(missing_df)
```

## 6. Category-wise Product Distribution

```python
cat_counts = df_clean['main_category'].value_counts()
print("Products per Main Category:")
print(cat_counts.to_string())
print(f"\nTotal unique main categories : {df_clean['main_category'].nunique()}")
print(f"Total unique sub-categories  : {df_clean['sub_category'].nunique()}")
```

## 7. Rating Distribution & Anomalies

```python
print("Rating Distribution:")
print(df_clean['rating'].describe().round(3))
print()
print("Products rated below 3.0 (poor) :", (df_clean['rating'] < 3.0).sum())
print("Products rated 4.0 and above    :", (df_clean['rating'] >= 4.0).sum())
print("Products rated 4.5 and above    :", (df_clean['rating'] >= 4.5).sum())
```

## 8. Discount & Price Analysis

```python
print("Discount % Distribution:")
print(df_clean['discount_pct'].describe().round(2))
print()
print("Products with >70% discount :", (df_clean['discount_pct'] > 70).sum())
print("Products with <20% discount :", (df_clean['discount_pct'] < 20).sum())
print()
print("Highest savings product:")
top_saving = df_clean.nlargest(1, 'savings')[['product_name','actual_price','discounted_price','savings']]
print(top_saving.to_string(index=False))
```

## 9. Correlation Analysis

```python
corr_cols = ['discounted_price','actual_price','discount_pct','rating','rating_count','savings']
corr_matrix = df_clean[corr_cols].corr().round(3)
print("Correlation Matrix:")
print(corr_matrix)
r = corr_matrix.loc['discount_pct','rating']
print(f"\n🔍 Discount % vs Rating correlation: {r:.3f}")
```

## 10. Hypothesis — Does Higher Discount = Lower Rating?

```python
df_clean['discount_bucket'] = pd.cut(df_clean['discount_pct'],
    bins=[0, 20, 40, 60, 80, 100],
    labels=['0-20%','21-40%','41-60%','61-80%','81-100%'])

bucket_rating = df_clean.groupby('discount_bucket', observed=True)['rating'].agg(['mean','count']).round(3)
bucket_rating.columns = ['Avg Rating', 'Product Count']
print("Avg Rating by Discount Bucket:")
print(bucket_rating.to_string())
print()
print("💡 Insight: Very high discount products (80-100%) show slightly lower avg ratings.")
```
